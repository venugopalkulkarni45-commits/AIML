```python
import pandas as pd
import matplotlib.pyplot as plt
count = df['variety'].value_counts()
count.plot(kind='bar',figsize=(5,4))
plt.title("variety count")
plt.xlabel("variety")
plt.ylabel("count")
plt.show()
```
