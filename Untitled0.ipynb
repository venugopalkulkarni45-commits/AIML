```python
import spacy
from nltk import ngrams
nlp = spacy.load("en_core_web_sm")
doc = nlp(" AI is simulation of humn intelligence by machine.It includes expects NLP speech recognition and machine vision ")
print("Bigrams :- ",list(ngrams(doc,2)))
print("Trigrams :- ",list(ngrams(doc,3)))
```
